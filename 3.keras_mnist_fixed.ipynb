# Keras MNIST

使用 Keras 建立多層感知器（MLP）辨識 MNIST 手寫數字。

```python
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense, Flatten

#載入 MNIST 手寫數字數據集
mnist = keras.datasets.mnist
(train_data, train_label), (test_data, test_label) = mnist.load_data()

#輸出訓練集與測試集的維度與類型
print(type(train_data))
print(train_data.shape)
print(type(train_label))
print(train_label.shape)
print(test_data.shape)
print(test_label.shape)

#輸出第一筆標籤與其圖片像素矩陣
print(train_label[0])
print(train_data[0])

#繪製並顯示第一張手寫數字影像
import matplotlib.pyplot as plt
plt.imshow(train_data[0], cmap='binary')
plt.show()

#建立簡單的 MLP 順序模型
model = Sequential() # 建立線性堆疊模型
model.add(Flatten()) # 將 28x28 的二維影像壓平成 784 維的一維向量
model.add(Dense(512, activation='relu', input_dim=784)) # 隱藏層包含 512 個神經元，使用 ReLU
model.add(Dense(10, activation='softmax')) # 輸出層包含 10 個神經元（對應數字 0-9），使用 Softmax 得到機率分佈

#編譯模型：設定優化器、多分類損失函數以及評估指標
model.compile(optimizer='rmsprop', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

#訓練模型：執行 20 個週期，每次批量處理大小為 512
model.fit(train_data, train_label, epochs=20, batch_size=512)

#在測試集上評估模型表現並輸出準確度
score = model.evaluate(test_data, test_label)
print("測試集準確度:", score[1])
```
